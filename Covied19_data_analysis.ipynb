#Q3 In which region,maximum no. of confirmed cases were recorded

```python
df.groupby('Region').Confirmed.sum().sort_values(ascending=False)
```

#Q4 show the region with the minimum no. of death cases

```python
df.groupby('Region').Deaths.sum().sort_values(ascending=True)
```
